# What is OOP?

Let's walk through the fundamentals of object-oriented programming, or OOP!

# Procedural vs OOP

Until now, we have likely been coding in procedural style: our code was a sequence of steps to be carried out. This is common when performing simple, non-repetitive tasks, or for data analysis, where we load data, explore it, and produce insights.

# Thinking in sequences

Procedural thinking is natural and how we tend to plan our days. But a software engineer, like a city planner, needs to think about thousands of people with their own routines. Mapping out a sequence of actions for each individual would become unsustainable. Instead, we should think about patterns of behavior.

# Procedural vs OOP

This is what OOP involves - thinking about objects and patterns of their interactions - like users interacting with elements of an interface. This point of view becomes invaluable when designing and building software like Python packages! It helps us organize our code better, making it more reusable and maintainable.

# Objects

The first fundamental concept of OOP is an object, which is a combination of data and functionality. For example, an object representing a customer can have data such as a phone number and email associated with them, and functionality like the ability to place or cancel an order. Data associated with an object is known as the object's state. The functionality of an object is known as its behavior. In OOP, state and behavior are bundled together. So, instead of thinking about customer data separately from customer actions, we think of them as one unit representing a customer.

# Objects in Python

In Python, everything is an object. Numbers, strings, DataFrames, and even functions are objects.

# Classes as blueprints

But how do objects get created? This leads to the other core concept of OOP, classes. Classes are like blueprints for objects. They describe the possible states and behaviors that every object from that class can have. Therefore, Customer objects are made from a Customer class. To enable Customer objects to have states such as email address and phone number and behaviors including placing and canceling orders, they must be defined in the Customer class. The class allows us to talk about customers in a unified way.

# Classes as blueprints

An object is an instance of a class. So an individual customer is an object created from our Customer class blueprint.

# What is self?

So what was that self? Classes are templates. So when we make the class there are not yet any associated objects. So, we include self as the first argument to methods as a way of referring to the object that we will create. Put another way, it's a stand-in for the future object. When a method is called from an object, it is the equivalent of calling the method from the class and passing the object, so cust.identify is the same as Customer.identify, where cust is passed as the first argument. That's why we don't specify it explicitly when calling the method from an existing object.

# Constructor

A better strategy would be to add data to the object when creating it, like we do when creating a list or DataFrame. Python allows us to do this by adding a special method to our class called the constructor. The constructor is automatically called every time an object is created. The method has to be called underscore underscore init underscore underscore - the exact name and double underscores are essential for Python to recognize it. Here, we define the init method for the Customer class. The method takes on one argument, name, in addition to the self argument that should be there for any method. In the body of the method, we create the name attribute using self.attribute syntax, assigning its value to the name parameter, and print a message.

```python
# Define Customer class
class Customer:

    # Define __init__() constructor w/ 2 args: self, name
    def __init__(self, name):

        # Assign name to .name attribute of self
        self.name = name

        # Call print() function w/ string as input
        print("The __init__ method was called")
```

 Now we can pass the customer name in parentheses when creating a customer object, resulting in the init method being automatically called and the name attribute assigned.

```python
# Call Customer() class w/ string as input: cust
cust = Customer("Lara de Silva")

# Call print() function w/ .name attribute of cust as input
print(cust.name)
```

So, there are two ways to define attributes: we can define an attribute in any method in a class; and then calling the method will add the attribute to the object. 

```python
# Call show_image() function w/ 2 args:
# 1st arg is string
# 2nd arg sets size to tuple of 2 floats/ints
show_image("attributes_in_methods_0", size=(50, 5))
```

```python
# Call show_image() function w/ 2 args:
# 1st arg is string
# 2nd arg sets size to tuple of 2 floats/ints
show_image("attributes_in_constructor_0", size=(25, 2.5))
```

Where possible, we should avoid defining attributes outside the constructor. Our class definition can be hundreds lines of code long, and the person reading it would have to comb through all of them to find all the attributes. Additionally, defining all attributes in the constructor ensures that all of them are created when the object is created, so we don't have to worry about trying to access an attribute that doesn't exist yet. All this results in more organized, readable, and maintainable code.

# Add arguments

We can add another parameter, account balance, to the init method, and create another attribute, also called balance, that will be initialized during object creation. 

```python
# Define Customer class
class Customer:

    # Define __init__() constructor w/ 3 args: self, name, balance
    def __init__(self, name, balance):

        # Assign name to .name attribute of self
        self.name = name

        # Assign name to .balance attribute of self
        self.balance = balance

        # Call print() function w/ string as input
        print("The __init__ method was called")
```

# Add a class constructor

In this exercise, you'll continue working on the Employee class. Instead of using the methods like set_salary() that you wrote in the previous lesson, you will introduce a constructor that assigns name and salary to the employee at the moment when the object is created. Initializing attributes in the constructor is a good idea because this ensures that the object has all the necessary attributes the moment it is created.  Let's start by defining the Employee class with a constructor that accepts two arguments, name and salary (the latter with a default value of 0).  We'll also create 2 attributes, also called name and salary, setting their values to the corresponding arguments.  Finally, we'll also create 2 methods that set 2 additional attribute of the class: give_raise() and monthly_salary.

```python
# Define Employee class
class Employee:

    # Define __init__() constructor w/ 3 args: self, name, salary
    # The salary default arg is 0    
    def __init__(self, name, salary=0):
    
        # Assign name to .name attribute of self    
        self.name = name
    
        # Assign salary to .salary attribute of self        
        self.salary = salary

    # Define give_raise() method w/ 2 args:
    # 1st arg is self
    # 2nd arg is amount 
    def give_raise(self, amount):

        # Increase .salary attribute of self by amount
        self.salary += amount

    # Define monthly_salary() method w/ self as input 
    def monthly_salary(self):

        # Return .salary method of self divided by 12
        return self.salary / 12 
```

Let's modify the init constructor to check whether the salary parameter is positive, assigning to the salary attribute if true, or assigning 0 and printing "Invalid salary!" if false.

```python
# Define Employee class
class Employee:
    
    # Define __init__() constructor w/ 3 args: self, name, salary
    # The salary default arg is 0
    def __init__(self, name, salary=0):
        
        # Assign name to .name attribute of self
        self.name = name
        
        # If salary is greater than 0...
        if salary > 0:
            
            # Assign salary to .salary attribute of self
            self.salary = salary
        
        # Otherwise...
        else:
            
            # Assign 0 to .salary attribute of self
            self.salary = 0
            
            # Call print() function w/ string as input
            print("Invalid salary!")
    
    # Define give_raise() method w/ 2 args:
    # 1st arg is self
    # 2nd arg is amount 
    def give_raise(self, amount):
        
        # Increase .salary attribute of self by amount        
        self.salary += amount
    
    # Define monthly_salary() method w/ self as input 
    def monthly_salary(self):
        
        # Return .salary method of self divided by 12        
        return self.salary / 12
```

Now let's create an object using our Employee class and print off the name and salary of the object.  

```python
# Call Employee() class w/ string as input: emp
emp = Employee("Korel Rossi")

# Call print() function w/ .name attribute of emp as input
print(emp.name)

# Call print() function w/ .salary attribute of emp as input
print(emp.salary)    
```

You're doing great! Conditional statements are ideal for validating class attributes, ensuring only appropriate values can be assigned.

# Building a class from scratch

In the last exercise of the chapter, you'll step into the role of a Python developer and build your own class from scratch including a constructor! You'll design a Calculator class, including methods that enable the addition, subtraction, or multiplication of two values.

```python
# Define Calculator class
class Calculator:
    
    # Define __init__() constructor w/ 3 args: self, num_one, num_two
    def __init__(self, num_one, num_two):
        
        # Assign num_one to .num_one attribute of self
        self.num_one = num_one

        # Assign num_two to .num_two attribute of self
        self.num_two = num_two

    # Define addition() method w/ self as input
    def addition(self):

        # Return sum of .num_one attribute of self & .num_two attribute of self
        return self.num_one + self.num_two

    # Define subtraction() method w/ self as input
    def subtraction(self):

        # Return .num_two attribute of self subtracted from .num_one attribute of self
        return self.num_one - self.num_two

    # Define multiplication() method w/ self as input
    def multiplication(self):

        # Return product of .num_one attribute of self & .num_two attribute of self
        return self.num_one * self.num_two
```

# Alternative constructors

So why would we ever need class methods? The main use case is to enable alternative constructors. A class can only have one init method, but there might be multiple ways to initialize an object. For example, we might want to create an Employee object from data stored in a file. We can't use a method, because it would require an instance, and there isn't one yet! Here, we introduce a class method from_file that accepts a file name, reads the file that contains the name of the employee in the first line and the salary in the second line, returning an object. In the return statement, we use the cls variable. Remember that cls refers to the class, so this will call the init constructor, just like using Employee with parentheses would when used outside the class definition. 

```python
# Define Employee class
class Employee:
    
    # Define __init__() constructor w/ 3 args: self, name, salary
    # The salary default arg is 0
    def __init__(self, name, salary):
        
        # Assign name to .name attribute of self
        self.name = name
           
        # Assign 0 to .salary attribute of self
        self.salary = salary

    # Call @classmethod decorator
    @classmethod

    # Define from_file() function w/ 2 args: cls, filename
    def from_file(cls, filename):

        # With open() function w/ 2 args as input as f...
        # 1st arg is filename
        # 2nd arg is string
        with open(filename, "r") as f:

            # Call .strip() method of .readline() method of f: name
            name = f.readline().strip()

            # Call int() function w/ .strip() method of .readline() method of f as input 
            salary = int(f.readline().strip())

        # Return clls() function w/ 2 inputs: name, salary
        return cls(name, salary)
```

```python
# Define Person class
class Person:

    # Create int: CURRENT_YEAR    
    CURRENT_YEAR = 2024

    # Define __init__() constructor w/ 3 args: self, name, age
    def __init__(self, name, age):
        
        # Assign name to .name attribute of self       
        self.name = name

        # Assign age to .age attribute of self
        self.age = age
    
    # Call @classmethod decorator
    @classmethod  

    # Define from_birth_year() function w/ 3 inputs: cls, name, birth_year
    def from_birth_year(cls, name, birth_year):

        # Subtract birth_year from .CURRENT_YEAR attribute of Person: age
        age = Person.CURRENT_YEAR - birth_year

        # Return cls() function w/ 2 args: name, age
        return cls(name, age)
```

# Building a BetterDate Class

You are developing a time series package and want to define your own class for working with dates, BetterDate. The attributes of the class will be year, month, and day. You want to have a constructor that creates BetterDate objects given the values for year, month, and day, but you also want to be able to create BetterDate objects from strings, such as 2021-04-30.

```python
# Define BetterDate class
class BetterDate:

    # Define __init__() constructor w/ 3 args: self, name, age
    def __init__(self, year, month, day):

        # Assign year to .year attribute of self               
        self.year = year

        # Assign month to .month attribute of self              
        self.month = month

        # Assign day to .day attribute of self               
        self.day = day
    
    # Call @classmethod decorator
    @classmethod  

    # Define from_str() function w/ 2 inputs: cls, date_str    
    def from_str(cls, datestr):

        # Call .split() method of datestr w/ string as input: parts
        parts = datestr.split("-")

        # Call int() functions w/ index 0 of parts as input: year
        year = int(parts[0]) 

        # Call int() functions w/ index 1 of parts as input: month
        month = int(parts[1]) 
        
        # Call int() functions w/ index 2 of parts as input: day
        day = int(parts[2]) 

        # Return cls() function w/ 3 inputs: year, month, day
        return cls(year, month, day)
```

# Implementing class inheritance

We have our BankAccount class here, so how do we implement a SavingsAccount class that inherits from BankAccount? 

```python
# Define BankAccount class        
class BankAccount:

    # Define __init__() constructor w/ 2 args: self, balance
    def __init__(self, balance):

        # Assign balance to .balance attribute of self
        self.balance = balance

    # Define withdraw() method w/ 2 args
    def withdraw(self, amount):

        # Subtract amount from .balance attribute of self
        self.balance -= amount
```

We declare a new class, but this time, we add parentheses after the class name and, inside, specify the class to inherit from. Using the pass keyword results in the SavingsAccount class having all the functionality of BankAccount. In inheritance, the class we inherit from is known as the Parent class, while the class that inherits from another is known as the Child class or subclass.

```python
# Define SavingsAccount() class w/ parent BankAccount class        
class SavingsAccount(BankAccount):

    # Pass
    pass
```

# Adding functionality

We can add methods to a child class in the same way we add them to any other class. Within the methods we add, we can use data from both the child and parent classes. For example, here, we add a compute_interest method that returns the amount of interest in the account. We include the n_periods argument. The method performs a calculation using n_periods, the interest_rate attribute from the SavingsAccount class, which is the child class, along with using self to access the balance attribute from the BankAccount class, which is the parent.

```python
# Define SavingsAccount() class w/ parent BankAccount class        
class SavingsAccount(BankAccount):

    # Define __init__() constructor w/ 3 args: self, balance, interest_rate
    def __init__(self, balance, interest_rate):

        # Call .__init__() method of BankAccount w/ 2 args: self, balance
        BankAccount.__init__(self, balance)

        # Assign interest_rate to .interest_rate attribute of self
        self.interest_rate = interest_rate

    # Define compute_interest() method w/ 2 args:
    # 1st arg is self
    # 2nd arg set n_periods to 1
    def compute_interest(self, n_periods=1):

        # Call .balance attribute of self and multiply by 1 plus .interest_rate attribute of self to the exponent of n_periods-1
        interest = self.balance * ((1+self.interest_rate)**n_periods-1)

        # Return interest
        return interest 
```

# Adding a second child class

To create CheckingAccount, we add code that executes the parent class BankAccount constructor. We follow this with a new deposit method, and a withdraw method, but we add a new argument to withdraw - a fee that represents the additional withdrawal fee. We compare the amount to the withdrawal limit from the constructor, and, if the condition is met, call the parent's withdraw method, subtracting the amount and the fee from the balance. If the amount is above the limit, the withdrawal is declined. So this method runs almost the same code as the BankAccount's withdraw method without re-implementing it - just augmenting.

```python
# Define CheckingAccount() class w/ parent BankAccount class        
class CheckingAccount(BankAccount):

    # Define __init__() constructor w/ 3 args: self, balance, limit
    def __init__(self, balance, limit):

        # Call .__init__() method of BankAccount w/ 2 args: self, balance
        BankAccount.__init__(self, balance)

        # Assign limit to .limit attribute of self
        self.limit = limit

    # Define deposit() method w/ 2 args:
    # 1st arg is self
    # 2nd arg is amount
    def deposit(self, amount):

        # Increase .balance attribute of self by amount
        self.balance += amount

    # Define withdraw() method w/ 3 args:
    # 1st arg is self
    # 2nd arg is amount
    # 3rd arg is fee set to 0
    def withdraw(self, amount, fee=0):

        # If amount is less than or equal to .limit attribute of self...
        if amount <= self.limit:

            # Call .withdraw() method of BankAccount class w/ 2 args:
            # 1st arg is self
            # 2nd arg is amount plus fee
            BankAccount.withdraw(self, amount+fee)

        # Otherwise...
        else:

            # Pass
            pass
```

# Method inheritance

Let's extend the Manager class (which is inherited from the Employee class), created in the previous exercise, by creating a method called give_raise(). It will be similar to the method of the same name in the Employee class, but include an additional argument called bonus.

```python
# Create Manager() class w/ input of Employee class
class Manager(Employee):

    # Define .__init__() method w/ 4 args:
    # 1st arg is self
    # 2nd arg is name
    # 3rd arg is salary w/ default of 50k
    # 4th arg is project w/ default of None
    def __init__(self, name, salary=50000, project=None):

        # Call .__init__() method of Employee w/ 3 args:
        # 1st arg is self
        # 2nd arg is name
        # 3rd arg is salary
        Employee.__init__(self, name, salary)

        # Set .project attribute of self to project
        self.project = project

    # Define .display() method w/ self as input
    def display(self):

        # Call print() function w/ 2 args:
        # 1st arg is string
        # 2nd arg is .name attribute of self
        print("Manager ", self.name)

    # Define give_raise() method w/ 2 args: 
    # 1st arg is self
    # 2nd arg is amount
    def give_raise(self, amount, bonus=1.05):

        # Multiply amount by bonus: new_amount
        new_amount = amount*bonus
        
        # Increase .salary attribute of Employee by new_amount
        Employee.give_raise(self, new_amount)

# Call Manager() class w/ 2 args: mngr
# 1st arg is string
# 2nd arg is int
mngr = Manager("Ashta Dunbar", 78500)

# Call .give_raise() method of mngr w 2 args
# 1st arg is int
# 2nd arg sets bonus to 1.03
mngr.give_raise(2000, bonus=1.03)

# Call .salary attribute of mngr
mngr.salary
```

# Object equality

As a class is a blueprint, we might expect two objects with identical attributes to be treated as equals, right? If we check that these two objects are equal, the result is "False". This makes sense: we can have two different customers with the same name and account balance.

Why? Python doesn't consider two objects with the same data equal by default because of how the objects and variables representing them are stored.

# Variables are references

If we try to print the value of a customer object, we'll see "Customer at" and a series of characters. When an object is created, Python allocates a chunk of memory to that object, and the variable the object is assigned to just contains the reference to the memory chunk. These characters represent the memory location. So, this code tells Python to allocate a chunk of memory for a customer object and label it customer_one, then allocate another chunk for customer_two. When we compare customer_one and customer_two, we are comparing references, not the data. Because the two objects point to different chunks in memory, they are not considered equal.

```python
# Call print() function w/ input of customer1
print(customer1)

# Call print() function w/ input of customer2
print(customer2)
```

# The __eq__() method

We can define a special method for this. We know about the __init__ constructor, which is implicitly called when an object is created. Similarly, there is the __eq__ method, which Python implicitly calls whenever two objects of the same class are compared. The method should accept two arguments, referring to the objects to be compared. They are usually called self and other by convention. It should always return a Boolean value - True or False. Here, we have a Customer class with id and name attributes, and we define our custom eq method passing self and other. We customize the return statement which, by default, looks at whether the two objects are allocated to the same memory chunk. It will now compare the id and name of the two objects, returning True if the values of these attributes are equal. We also included a printout for illustration.

```python
# Create Customer class
class Customer:

    # Define .__init__() method w/ 3 args: self, acc_id, name
    def __init__(self, acc_id, name):

        # Assign acc_id, name to .acc_id, .name atrributes of self
        self.acc_id, self.name = acc_id, name

    # Define .__eq__() method w/ 2 args: self, other
    def __eq__(self, other):

        # Call print() function w/ string as input
        print("__eq__() is called")

        # Create boolean: acc_id_name_check_bool
        # Check if .acc_id attribute of self & other are the same
        # Check if .name attribute of self and .name attribute of other are the same
        acc_id_name_check_bool = (self.acc_id == other.acc_id) and (self.name==other.name)
        
        # Return acc_id_name_check_bool
        return acc_id_name_check_bool
```

# Checking types

This is a good start, but what if we tried comparing two objects from different classes with the same attributes and values? Python will see them as equal! To avoid this, we also need to ensure that the two objects are from the same class, which we can do by checking their type within our eq method.

```python
# Create Customer class
class Customer:

    # Define .__init__() method w/ 3 args: self, acc_id, name
    def __init__(self, acc_id, name):

        # Assign acc_id, name to .acc_id, .name atrributes of self
        self.acc_id, self.name = acc_id, name

    # Define .__eq__() method w/ 2 args: self, other
    def __eq__(self, other):

        # Create boolean: bool_check
        # Check if .acc_id attribute of self & other are the same
        # Check if .name attribute of self and .name attribute of other are the same
        # Check if type() function w/ self as input is equal to type() function w/ other as input
        bool_check = (self.acc_id == other.acc_id) and (self.name==other.name) and (type(self) == type(other))
        
        # Return bool_check
        return bool_check
```

# Overloading equality

Below is the BankAccount class. It has two attributes, balance and number, and a withdraw() method. Two bank accounts with the same balance are not necessarily the same account, but a bank account usually has an account number, and two accounts with the same account number should be considered the same.  Let's ensure that they are using the .__eq__() method within the class.  

```python
# Define BankAccount() class
class BankAccount():

    # Define __init__() constructor w/ 3 args
    # 1st arg is self
    # 2nd arg is number
    # 3rd arg is balance set to 0 
    def __init__(self, number, balance=0):

        # Assign balance to .balance attribute of self
        self.balance = balance

        # Assign number to .number attribute of self
        self.number = number

    # Define .withdraw() method w/ 2 args: self, amount
    def withdraw(self, amount):

        # Subtract amount from .balance attribute of self
        self.balance -= amount 
        
    # Define .__eq__() method w/ 2 args: self, other 
    def __eq__(self, other):
        
        # Return boolean of if .number attribute of self equals .number attribute of other
        return self.number == other.number   

# Call BankAccount() class w/ 2 ints as args: acct1
acct1 = BankAccount(123, 1000)

# Call BankAccount() class w/ 2 ints as args: acct2
acct2 = BankAccount(123, 1000)

# Call BankAccount() class w/ 2 ints as args: acct3
acct3 = BankAccount(456, 1000)

# Call print() function w/ input of boolean of whether acct1 equals acct2
print(acct1 == acct2)

# Call print() function w/ input of boolean of whether acct1 equals acct2
print(acct1 == acct3)
```

# Checking class equality

In the previous exercise, you defined a BankAccount class with a number attribute that was used for comparison. But if you were to compare a BankAccount object to an object of another class that also has a number attribute, you could end up with unexpected results.  Let's modify the .__eq__ method to test whether the type of self and other are the same.  

```python
# Define BankAccount() class        
class BankAccount():

    # Define __init__() constructor w/ 3 args
    # 1st arg is self
    # 2nd arg is number
    # 3rd arg is balance set to 0 
    def __init__(self, number, balance=0):

        # Assign balance to .balance attribute of self
        self.balance = balance

        # Assign number to .number attribute of self
        self.number = number

    # Define .withdraw() method w/ 2 args: self, amount
    def withdraw(self, amount):

        # Subtract amount from .balance attribute of self
        self.balance -= amount 
        
    # Define .__eq__() method w/ 2 args: self, other 
    def __eq__(self, other):
        
        # Return boolean of if ...
        # .number attribute of self equals .number attribute of other
        # type() function w/ self as input equals type() function w/ other as input
        return (self.number == other.number) and (type(self) == type(other))   
```

Now let's create a Phone class that also has a .number attribute.  

```python
# Create Phone class
class Phone:

    # Define .__init__() method w/ self, number as inputs
    def __init__(self, number):

        # Assing number to .number attribute of self
        self.number = number

    # Define .__eq__() method w/ self, other as inputs
    def __eq__(self, other):

        # Return boolean of if
        # .number attribute of self equals .number attribute of other
        # type() function w/ self as input equals type() function w/ other as input
        return (self.number == other.number)
```

Perfect! Now, only comparing objects of the same class, BankAccount, could return True. Another way to ensure that an object has the same type as you expect is to use the isinstance(obj, Class) function. This can be helpful when handling inheritance, as Python considers an object to be an instance of both the parent and the child class. Try running pn == acct (with reversed order of equality). What does this tell you about the __eq__() method?

```python
# Call print() function w/ input of boolean of whether pn equals acct
print(pn == acct)
```

With custom __eq__ methods, x == y calls the left operand's method and only falls back to the right operand's if the first returns NotImplemented.  When acct is the left operand, the BankAccount() class's .__eq__() method decides the boolean.  The BankAccount() class's .__eq__() method includes a type() check so we get a False value.  However, when pn is the left operand, the Phone() class's .__eq__() method decides the boolean.  The Phone() class's .__eq__() method does not include a type() check so we get a True even though two different classes are being compared.

```python
# Define BankAccount() class        
class BankAccount():

    # Define __init__() constructor w/ 3 args
    # 1st arg is self
    # 2nd arg is number
    # 3rd arg is balance set to 0 
    def __init__(self, number, balance=0):

        # Assign balance to .balance attribute of self
        self.balance = balance

        # Assign number to .number attribute of self
        self.number = number

    # Define .withdraw() method w/ 2 args: self, amount
    def withdraw(self, amount):

        # Subtract amount from .balance attribute of self
        self.balance -= amount 
        
    # Define .__eq__() method w/ 2 args: self, other 
    def __eq__(self, other):

        # Call print() funciton w/ string as input
        print("BankAccount __eq__() called")
        
        # Return boolean of if .number attribute of self equals .number attribute of other
        return (self.number == other.number)

# Define SavingsAccount() class w/ parent BankAccount class        
class SavingsAccount(BankAccount):

    # Define __init__() constructor w/ 4 args: self, number, balance, interest_rate
    def __init__(self, number, balance, interest_rate):

        # Call .__init__() method of BankAccount w/ 3 args: self, number, balance
        BankAccount.__init__(self, number, balance)

        # Assign interest_rate to .interest_rate attribute of self
        self.interest_rate = interest_rate

    # Define .__eq__() method w/ 2 args: self, other 
    def __eq__(self, other):

        # Call print() funciton w/ string as input
        print("SavingsAccount __eq() called")
        
        # Return boolean of if .number attribute of self equals .number attribute of other
        return (self.number == other.number) 
```

# str vs repr

There are two special methods that we can define in a class that will return a printable representation of an object. The __str__ method is executed when we call print or str on an object, and the repr method is executed when we call repr on the object or print it in the console without explicitly calling print. 

```python
# Call show_image() function w/ 2 args:
# 1st arg is string
# 2nd arg sets size to tuple of 2 floats/ints
show_image("str_vs_repr_0", size=(30, 3.0))
```

The difference is that str gives an informal representation that is suitable for an end user, and repr is mainly used by developers. The best practice is to use repr to print a string that can be used to reproduce the object. For example, our NumPy array shows the exact method call that was used to create the object. If we only implement one of them, we should choose repr because it is also used as a fall-back for print when str is not defined.

# Implementation: repr

Starting with the repr method, we can define it to accept one argument, self, and return a string. The return statement uses an f-string, which is where we write "f" right before our quotes. This technique allows the insertion of variables inside curly brackets. Here, we provide self.name and self.balance. We include single quotes around self.name to ensure that repr returns the code that can be used to reproduce the object - the name as a string and balance as a number. 

```python
# Create Customer class
class Customer:

    # Define .__init__() method w/ 3 args: self, name, balance
    def __init__(self, name, balance):

        # Assign name to .name atrribute of self
        # Assign balance to .balance attribute of self 
        self.name, self.balance = name, balance

    # Define .__repr__() method w/ self as input
    def __repr__(self):

        # Return f-string w/ .name & .balance attribute of self
        return f"Customer('{self.name}', {self.balance})"
```

We can create an object and then display it, which shows the output of repr. 

```python
# Call Customer() class w/ string & int as args: cust
cust = Customer("Maryam Azar", 3000)

# Call cust
cust
```

As this class doesn't have the str method, repr will be used if we call the print method on the object as well.

```python
# Call print() function w/ cust as input
print(cust)
```

# Implementation: str

The str method also shouldn't accept any arguments besides self, and it should return a string. Here, the string representation of a customer will consist of the word Customer, then on the next line, name, colon, followed by the customer's name, then balance, colon, and the customer's balance. We use triple quotes to define multi-line strings in this case and again use an f-string to provide variables. 

```python
# Create Customer class
class Customer:

    # Define .__init__() method w/ 3 args: self, name, balance
    def __init__(self, name, balance):

        # Assign name to .name atrribute of self
        # Assign balance to .balance attribute of self 
        self.name, self.balance = name, balance

    # Define .__str__() method w/ self as input
    def __str__(self):

        # Create F-string w/ .name & .balance attribute of self: cust_str
        cust_str = f"""Customer Information:\nName: {self.name}\nBalance: {self.balance}"""
        
        # Return cust_str
        return cust_str
```

# String representation of objects

There are two special methods in Python that return a string representation of an object. The __str__() method is called when you use print() or str() on an object, and __repr__() is called when you use repr() on an object, print the object in the console without calling print(), or instead of __str__() if __str__() is not defined. The __str__() method is supposed to provide a "user-friendly" output describing an object, and __repr__() should return the expression that, when evaluated, will return the same object, ensuring the reproducibility of your code.  Let's continue working with the Employee class.  We'll start by creating the __repr__() method within the Employee class.

```python
# Create Employee class
class Employee:

    # Define .__init__() method w/ 3 args:
    # 1st arg is self
    # 2nd arg is name
    # 3rg arg is salary set to default of 30k
    def __init__(self, name, salary=30000):

        # Set .name attribute of self to name 
        self.name = name

        # Set .salary attribute of self to salary 
        self.salary = salary

    # Define .__repr__() method w/ self as input
    def __repr__(self):

        # Create f-string w/ .name & .salary attributes of self
        emp_str = f"Employee('{self.name}', {self.salary})"

        # Return emp_str
        return emp_str
```

Now let's create 2 employees and see what printing the objects prodouces.

```python
# Create Employee() object w string & int as inputs: emp1
emp1 = Employee("Amar Howard", 30000)

# Call print() function w/ input of emp2
print(emp1)

# Create Employee() object w string & int as inputs: emp2
emp2 = Employee("Carolyn Ramirez", 35000)

# Call print() function w/ input of emp2
print(emp2)
```

Because we defined the __repr__ method when creating the object, but did NOT define the __str__ method, printing the objects gave us the string that was used to create the object.  This is because when we defined the __repr__ method above, we created said F-string.  Now let's see what happens when we create both the __repr__ and __str__ methods for the object.  

```python
# Create Employee class
class Employee:

    # Define .__init__() method w/ 3 args:
    # 1st arg is self
    # 2nd arg is name
    # 3rg arg is salary set to default of 30k
    def __init__(self, name, salary=30000):

        # Set .name attribute of self to name 
        self.name = name

        # Set .salary attribute of self to salary 
        self.salary = salary
    
    # Define .__repr__() method w/ self as input
    def __repr__(self):

        # Create f-string w/ .name & .salary attributes of self
        emp_str = f"Employee('{self.name}', {self.salary})"

        # Return emp_str
        return emp_str

    # Define .__str__() method w/ self as input
    def __str__(self):

        # Create F-string w/ .name & .salary attribute of self: emp_str
        emp_str = f"""Employee name: {self.name}\nEmployee salary: {self.salary}"""
        
        # Return emp_str
        return emp_str

# Create Employee() object w string & int as inputs: emp1
emp1 = Employee("Amar Howard", 30000)

# Call print() function w/ input of emp1
print(emp1)

# Create Employee() object w string & int as inputs: emp2
emp2 = Employee("Carolyn Ramirez", 35000)

# Call print() function w/ input of emp2
print(emp2)
```

Now printing the objects outputs a more user friendly F-string because the __str__ method was created.  When both the __repr__ and the __str__ method exist for a class, using print() function will default to the user friendly __str__ method's output.  If we want to access either __str__ or __repr__, we can also just call them directly as functions.   

```python
# Create Employee() object w string & int as inputs: emp1
emp1 = Employee("Amar Howard", 30000)

# Call print() function w/ str() function w/ input of emp1 as input
print(str(emp1))

# Call print() function w/ empty string as input
print("")

# Call print() function w/ repr() function w/ input of emp1 as input
print(repr(emp1))

# Call print() function w/ empty string as input
print("")
```

# Exception handling

For example, we might want to run the rest of the script but log the error without terminating the program. We can implement a special code to handle this case. We can use a try-except-finally sequence to catch an exception and handle it. Start with try, then place the code we're worried won't work inside. Let's demonstrate this with code we know will cause an error. Here, we try to add 5 and the string "a". Then, we add an except block, stating the name of the particular exception we want to handle, in this case a TypeError. Inside the except block, we provide the code that should be executed if this exception is raised, showing a print statement. Then, if this particular exception does happen, the program will not terminate but execute the code in the except block instead. We can also have multiple exception blocks. Lastly, we can use the optional "finally" block that will contain the code that runs no matter what.

```python
# Try...
try:

    # Call print() function w/ int combined w/ string as input
    print(5 + "a")

# If TyppError exception occurs...
except TypeError:

    # Call print() function w/ string as input
    print("You can't add an integer to a string, but you can multipy them!")

# If ValueError exception occurs...
except ValueError:

    # Call print() function w/ string as input
    print("You triggered a ValueError!")

# Finally...
finally:

    # Call print() function w/ int multiplied by string as input
    print(5 * "a")
```

# Exception handling output

# Raising exceptions

Sometimes, we want to raise exceptions ourselves, for example when some conditions aren't satisfied. We can use the raise keyword followed by the exception to raise, optionally adding a specific error message in parentheses. 

```python
# Define make_list_of_ones() function w/ input of length
def make_list_of_ones(length):

    # If length is less than or equal to 0...
    if length <= 0:

        # Raise ValueError() function w/ string as input
        raise ValueError("Invalid length!")

    # Return list of int multiplied by length
    return [1]*length
```

# Catching exceptions

The function invert_at_index(x, ind) takes two arguments, a list x and an index ind, and inverts the element of the list at that index. For example invert_at_index([5,6,7], 0) returns 1/5, or 0.2.  Within the function we also catch 2 different types of exceptions which will trigger print statements.

```python
# Define invert_at_index() function w/ x, ind as inputs
def invert_at_index(x, ind):

    # Try...
    try:

        # Return 1 divided by index ind of x
        return 1/x[ind]

    # If ZeroDivisionError exception is triggered...
    except ZeroDivisionError:

        # Call print() function w/ string as input
        print("Cannot divide by zero!")

    # If IndexError exception is triggered...
    except IndexError:

        # Call print() function w/ string as input
        print("Index out of range!")

```

# Custom exceptions

You don't have to rely solely on built-in exceptions like IndexError: you can define custom exceptions that are more specific to your application. You can also define exception hierarchies. All you need to define an exception is a class inherited from the built-in Exception class or one of its subclasses. Earlier in the course, you defined an Employee class and used print statements and default values to handle errors like creating an employee with a salary below the minimum or giving a raise that is too large. A better way to handle this situation is to use exceptions - because these errors are specific to our application (unlike, for example, a division by zero error, which is universal), it makes sense to use custom exception classes.  Let's start by creating two custom exceptions.  The first is the SalaryError class which inherits from Parent ValueError.  

```python
# Create SalaryError class with parent ValueError
class SalaryError(ValueError): 

    # Pass
    pass
```

Next we'll create a BonusError() class that inherits from SalaryError

```python
# Create BonusError class with parent SalaryError
class BonusError(SalaryError): 

    # Pass
    pass      
```

Now let's redefine our Employee class.  We'll create class level attributes MIN_SALARY and MAX_BONUS.  We'll also add in custom Error handling telling the user when salaries are too low or bonuses are too high.  

```python
# Create Employee class
class Employee:

    # Assign int to class level attribute: MIN_SALARY
    MIN_SALARY = 30000

    # Assign int to class level attribute: MAX_BONUS
    MAX_BONUS = 5000

    # Define .__init__() attribute w/ 3 args:
    # 1st arg is self
    # 2nd arg is name
    # 3rd arg sets salary to default of 30k
    def __init__(self, name, salary = 30000):

        # Assign name to .name attribute of self
        self.name = name    

        # If salary is less than .MIN_SALARY attribute of Employee...
        if salary < Employee.MIN_SALARY:

            # Raise SalaryError() w/ string as input
            raise SalaryError("Salary is too low!")

        # Assign salary to .salary attribute of self
        self.salary = salary

    # Define .give_bonus() method w/ self, amount as inputs  
    def give_bonus(self, amount):

        # If amount is greater than .MAX_BONUS attribute of Employee...
        if amount > Employee.MAX_BONUS:

            # Raise BonusError() w/ string as input
            raise BonusError("The bonus amount is too high!")  

        # Otherwise if .salary attribute of self plus amount is less than .MIN_SALARY attribute of Employee...
        elif self.salary + amount <  Employee.MIN_SALARY:

            # Raise SalaryError() w/ string as input
            raise SalaryError("The salary after bonus is too low!")

        # Increase .salary attribute of self by amount
        self.salary += amount
```
